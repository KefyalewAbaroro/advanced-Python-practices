```python
import matplotlib.pyplot as plt
import numpy as np

# Data
years = [2006 + x for x in range(16)]
weights = [80,85,87,88,82,80,89,80,85,86,83,87,79,84,78,84]

plt.figure(figsize=(10,10))

# Compute percentiles for each weight
percentiles = [np.round(np.percentile(weights, (w / max(weights)) * 100), 1) for w in weights]

# Build labels using percentiles
labels = [f"{y}\n{p}%" for y, p in zip(years, percentiles)]

# Use a different color for each sector
colors = plt.cm.tab20(np.linspace(0, 1, len(weights)))

# Intuitive pie chart
plt.pie(
    weights,
    labels=labels,
    autopct='%1.1f%%',        # percentage of total
    startangle=90,
    colors=colors,
    textprops={'fontsize': 12, 'fontweight': 'bold'}
)

plt.title("Pie Chart: Weight Percentile Distribution by Year", fontsize=18)
plt.show()

```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
years = [2006 + x for x in range(16)]
weights = [80,85,87,88,82,80,89,80,85,86,83,87,79,84,78,84]

plt.figure(figsize=(12,7))

# Scatter plot with intuitive styling
plt.scatter(
    years,
    weights,
    c=weights,                 # color based on value
    cmap='viridis',            # intuitive gradient
    s=[w*5 for w in weights],  # size based on weight
    edgecolor='black',
    linewidth=1.2
)

# Add labels for each point
for x, y in zip(years, weights):
    plt.text(
        x,
        y + 0.8,
        str(y),
        ha='center',
        fontsize=11,
        fontweight='bold',
        color='green'
    )

# Trend line (intuitive)
z = np.polyfit(years, weights, 1)
p = np.poly1d(z)
plt.plot(years, p(years), "r--", linewidth=2)

# Titles and labels
plt.title("Scatter Plot: Weight by Year", fontsize=20)
plt.xlabel("Year")
plt.ylabel("Weight (lb)")
plt.grid(True, linestyle='--', alpha=0.6)

plt.show()
```

```python
# Use the store_date to plot the store size for each region using a bar chart
import matplotlib.pyplot as plt
import numpy as np

# Data from store_data.csv
regions = ['Lisbon', 'Porto', 'Algarve', 'Online', 'Coimbra']
sizes = [179, 728, 336, 950, 238]

plt.figure(figsize=(12,7))

# Determine thresholds
min_size = min(sizes)
max_size = max(sizes)
median_size = np.median(sizes)

colors = []
for s in sizes:
    if s == max_size:
        colors.append('green')      # highest value
    elif s == min_size:
        colors.append('red')        # lowest value
    else:
        colors.append('orange')     # medium values

# Plot bars
bars = plt.bar(regions, sizes, color=colors, edgecolor='black', linewidth=2)

# Add green bold labels
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height + 10,
        str(height),
        ha='center',
        fontsize=12,
        fontweight='bold',
        color='green'
    )

# Titles and labels
plt.title("Store Size by Region (Bar Chart)", fontsize=20)
plt.xlabel("Region")
plt.ylabel("Store Size (m²)")
plt.grid(True, linestyle='--', alpha=0.6)

plt.show()

```

```python
# Plot the price attribute for the Airbnb data using a histogram
import matplotlib.pyplot as plt
import numpy as np

# Example Airbnb price data (replace with df_airbnb['price'])
import matplotlib.pyplot as plt
import numpy as np

# Example Airbnb price data (replace with your df_airbnb['price'])
prices = [80,120,150,200,300,450,500,600,700,800,900,1000,1200,1300,1500]

plt.figure(figsize=(12,7))

# Histogram
n, bins, patches = plt.hist(
    prices,
    bins=10,
    edgecolor='black',
    linewidth=1.2
)

# Determine thresholds
avg_price = np.mean(prices)
high_price = np.percentile(prices, 75)
low_price = np.percentile(prices, 25)
close_high = np.percentile(prices, 60)

# Assign colors based on ranges
for patch, bin_edge in zip(patches, bins[:-1]):
    if bin_edge >= high_price:
        patch.set_facecolor('red')        # high values
    elif bin_edge >= close_high:
        patch.set_facecolor('yellow')     # close to high
    elif bin_edge >= avg_price:
        patch.set_facecolor('green')      # average values
    elif bin_edge <= low_price:
        patch.set_facecolor('orange')     # low values
    else:
        patch.set_facecolor('lightblue')  # fallback color

# Add green bold labels on each bin
for count, bin_edge in zip(n, bins[:-1]):
    plt.text(
        bin_edge + (bins[1] - bins[0]) / 2,
        count + 0.5,
        f"{int(count)}",
        ha='center',
        fontsize=11,
        fontweight='bold',
        color='green'
    )

# Titles and labels
plt.title("Histogram: Airbnb Price Distribution", fontsize=20)
plt.xlabel("Price ($)")
plt.ylabel("Frequency")
plt.grid(True, linestyle='--', alpha=0.6)

plt.show()
```

```python
# Create a Python list and plot the elements using a scatter plot
import matplotlib.pyplot as plt
import numpy as np

# World Cup years
world_cups = [1990, 1994, 1998, 2002, 2006, 2010, 2014, 2018, 2022]

# Intervals between tournaments
intervals = [world_cups[i+1] - world_cups[i] for i in range(len(world_cups)-1)]

# X-axis positions
x = range(len(intervals))

plt.figure(figsize=(12,7))

# Different colors for each point
colors = plt.cm.tab10(np.linspace(0, 1, len(intervals)))

# Scatter plot
plt.scatter(
    x,
    intervals,
    c=colors,
    s=[v*50 for v in intervals],   # size based on interval value
    edgecolor='black',
    linewidth=1.2
)

# Add green bold labels for each point
for i, v in enumerate(intervals):
    plt.text(
        i,
        v + 0.1,
        f"{v} yrs",
        ha='center',
        fontsize=12,
        fontweight='bold',
        color='green'
    )

# Titles and labels
plt.title("Scatter Plot: World Cup Intervals", fontsize=20)
plt.xlabel("Tournament Index")
plt.ylabel("Interval (Years)")
plt.grid(True, linestyle='--', alpha=0.6)

plt.show()
```
