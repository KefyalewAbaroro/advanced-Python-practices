```python
# Outliers Detection Using Z-Score
import pandas as pd
import numpy as np
from scipy.stats import zscore

# Load dataset
df_airbnb = pd.read_csv("Airbnb.csv")

# Clean price column if needed
df_airbnb['price'] = (
    df_airbnb['price']
    .replace('[\$,]', '', regex=True)
    .astype(float)
)

# Compute z-scores
df_airbnb['z'] = zscore(df_airbnb['price'], nan_policy='omit')

# Identify outliers (|z| > 3)
outliers_z = df_airbnb.loc[df_airbnb['z'].abs() > 3].copy()

outliers_z
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load dataset
df_airbnb = pd.read_csv("Airbnb.csv")

# Clean price column if needed
df_airbnb['price'] = (
    df_airbnb['price']
    .replace('[\$,]', '', regex=True)
    .astype(float)
)

# Plot boxplot
df_airbnb['price'].plot.box()
plt.title("Airbnb Price Distribution")
plt.ylabel("Price")
plt.show()
```
