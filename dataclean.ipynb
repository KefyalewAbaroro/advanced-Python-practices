```python
import pandas as pd

store_df = pd.read_csv("store_data.csv")
sales_df = pd.read_csv("sales_data.csv")

merged = pd.merge(
    sales_df,
    store_df,
    on="store_id",
    how="left"
)

print(merged.head())      # See first rows
print(merged.shape)       # See size
merged.info()             # See structure
```

```python
# Quantity Sold by Region (Bar Chart)
import matplotlib.pyplot as plt

region_qty = merged.groupby("region")["quantity"].sum().sort_values()

plt.figure(figsize=(10,6))
region_qty.plot(kind="bar", color="skyblue")
plt.title("Quantity Sold by Region")
plt.xlabel("Region")
plt.ylabel("Total Quantity Sold")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Quantity Sold by Store (Horizontal Bar Chart)
store_qty = merged.groupby("store_name")["quantity"].sum().sort_values()

plt.figure(figsize=(10,8))
store_qty.plot(kind="barh", color="green")
plt.title("Quantity Sold by Store")
plt.xlabel("Total Quantity Sold")
plt.ylabel("Store")
plt.tight_layout()
plt.show()
```

```python
# Store Size vs Quantity Sold (Scatter Plot)
plt.figure(figsize=(8,6))
plt.scatter(merged["store_size_m2"], merged["quantity"], alpha=0.6)
plt.title("Store Size vs Quantity Sold")
plt.xlabel("Store Size (m²)")
plt.ylabel("Quantity Sold")
plt.tight_layout()
plt.show()
```

```python
# Daily Quantity Trend (Line Chart)
merged["date"] = pd.to_datetime(merged["date"])

daily_qty = merged.groupby("date")["quantity"].sum()

plt.figure(figsize=(12,6))
daily_qty.plot(kind="line", color="purple")
plt.title("Daily Quantity Trend")
plt.xlabel("Date")
plt.ylabel("Quantity Sold")
plt.tight_layout()
plt.show()
```

```python
# Categorize customers into young / middle age / seniors
def categorize_age(age):
    if age < 30:
        return "young"
    elif age < 60:
        return "middle age"
    else:
        return "seniors"

customer["age_group"] = customer["age"].apply(categorize_age)
```
