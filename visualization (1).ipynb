```python
# Use the store_date to plot the store size for each region using a bar chart
import matplotlib.pyplot as plt
import numpy as np

# Data from store_data.csv
regions = ['Lisbon', 'Porto', 'Algarve', 'Online', 'Coimbra']
sizes = [179, 728, 336, 950, 238]

plt.figure(figsize=(12,7))

# Determine thresholds
min_size = min(sizes)
max_size = max(sizes)
median_size = np.median(sizes)

colors = []
for s in sizes:
    if s == max_size:
        colors.append('green')      # highest value
    elif s == min_size:
        colors.append('red')        # lowest value
    else:
        colors.append('orange')     # medium values

# Plot bars
bars = plt.bar(regions, sizes, color=colors, edgecolor='black', linewidth=2)

# Add green bold labels
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height + 10,
        str(height),
        ha='center',
        fontsize=12,
        fontweight='bold',
        color='green'
    )

# Titles and labels
plt.title("Store Size by Region (Bar Chart)", fontsize=20)
plt.xlabel("Region")
plt.ylabel("Store Size (m²)")
plt.grid(True, linestyle='--', alpha=0.6)

plt.show()

```

```python
# Plot the price attribute for the Airbnb data using a histogram
import matplotlib.pyplot as plt
import numpy as np

# Example Airbnb price data (replace with df_airbnb['price'])
import matplotlib.pyplot as plt
import numpy as np

# Example Airbnb price data (replace with your df_airbnb['price'])
prices = [80,120,150,200,300,450,500,600,700,800,900,1000,1200,1300,1500]

plt.figure(figsize=(12,7))

# Histogram
n, bins, patches = plt.hist(
    prices,
    bins=10,
    edgecolor='black',
    linewidth=1.2
)

# Determine thresholds
avg_price = np.mean(prices)
high_price = np.percentile(prices, 75)
low_price = np.percentile(prices, 25)
close_high = np.percentile(prices, 60)

# Assign colors based on ranges
for patch, bin_edge in zip(patches, bins[:-1]):
    if bin_edge >= high_price:
        patch.set_facecolor('red')        # high values
    elif bin_edge >= close_high:
        patch.set_facecolor('yellow')     # close to high
    elif bin_edge >= avg_price:
        patch.set_facecolor('green')      # average values
    elif bin_edge <= low_price:
        patch.set_facecolor('orange')     # low values
    else:
        patch.set_facecolor('lightblue')  # fallback color

# Add green bold labels on each bin
for count, bin_edge in zip(n, bins[:-1]):
    plt.text(
        bin_edge + (bins[1] - bins[0]) / 2,
        count + 0.5,
        f"{int(count)}",
        ha='center',
        fontsize=11,
        fontweight='bold',
        color='green'
    )

# Titles and labels
plt.title("Histogram: Airbnb Price Distribution", fontsize=20)
plt.xlabel("Price ($)")
plt.ylabel("Frequency")
plt.grid(True, linestyle='--', alpha=0.6)

plt.show()
```

```python
# Create a Python list and plot the elements using a scatter plot
import matplotlib.pyplot as plt
import numpy as np

# All-time World Cup winners with every year they won
winners = [
    ("Uruguay", 1930),
    ("Italy", 1934),
    ("Italy", 1938),
    ("Uruguay", 1950),
    ("Germany", 1954),
    ("Brazil", 1958),
    ("Brazil", 1962),
    ("England", 1966),
    ("Brazil", 1970),
    ("Germany", 1974),
    ("Argentina", 1978),
    ("Italy", 1982),
    ("Argentina", 1986),
    ("Germany", 1990),
    ("Brazil", 1994),
    ("France", 1998),
    ("Brazil", 2002),
    ("Italy", 2006),
    ("Spain", 2010),
    ("Germany", 2014),
    ("France", 2018),
    ("Argentina", 2022)
]

# Extract countries and years
countries = [w[0] for w in winners]
years = [w[1] for w in winners]

# Unique countries
unique_countries = sorted(set(countries))

# Assign a fixed color to each country
color_map = {country: plt.cm.tab10(i / len(unique_countries)) 
             for i, country in enumerate(unique_countries)}

# Map each winner entry to its country color
colors = [color_map[country] for country in countries]

# X-axis positions
x = np.arange(len(winners))

plt.figure(figsize=(16,8))

# Scatter plot
plt.scatter(
    x,
    years,
    c=colors,
    s=[(2026 - y) * 0.6 for y in years],  # size based on recency
    edgecolor='black',
    linewidth=1.0
)

# Add labels for each point
for i, (country, year) in enumerate(winners):
    plt.text(
        i,
        year + 1.5,
        f"{country}\n({year})",
        ha='center',
        fontsize=14,
        fontweight='bold',
        color='green'
    )

# Titles and labels
plt.title("Scatter Plot: All-Time World Cup Winners (Every Winning Year)", fontsize=20)
plt.xlabel("Winning Instance Index")
plt.ylabel("Year Won")
plt.grid(True, linestyle='--', alpha=0.6)

plt.xticks(x, [f"{country} {year}" for country, year in winners], rotation=90)

plt.tight_layout()
plt.show()

```
